Lee en voz alta el texto de `Payikos.txt` con la voz de macOS (`say`). No es la voz de Anuel: es una voz de sistema en español.

![Anuel AA](WhatsApp-Image-2026-10-02-at-11.30.42-1-2-e1790965145151.jpeg)

```python
import subprocess
from pathlib import Path

voz = "Jorge"  # cambia a "Paulina", "Monica", etc.
ruta = Path("Payikos.txt")  # Jupyter usa la carpeta del notebook como directorio de trabajo
```

```python
# Solo la canción del principio: se corta donde empieza el dibujo ASCII (línea "___")
# y se saltan las etiquetas de artista, como "(Ñengo)", y los créditos.
texto = ruta.read_text(encoding="utf-8").split("\n___")[0]
ignorar = {"chorus", "Killatonez", "Juanka & Killatonez", "Ñengo Flow"}
lineas = [
    l.strip()
    for l in texto.splitlines()
    if l.strip() and not l.strip().startswith("(") and l.strip() not in ignorar
]
print(len(lineas), "líneas")
```

```python
subprocess.run(["say", "-v", voz, "-r", "170", ". ".join(lineas)], check=True)
```
