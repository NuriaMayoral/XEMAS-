Lee en voz alta el texto de `Payikos.txt` con la voz de macOS (`say`). No es la voz de Anuel: es una voz de sistema en español.

```python
import subprocess
from pathlib import Path

voz = "Jorge"  # cambia a "Paulina", "Monica", etc.
ruta = Path("Payikos.txt")  # Jupyter usa la carpeta del notebook como directorio de trabajo
```

```python
# Nos quedamos con las líneas que son sobre todo letras (se saltan los dibujos ASCII).
lineas = [
    l.strip()
    for l in ruta.read_text(encoding="utf-8").splitlines()
    if l.strip() and sum(c.isalpha() or c == " " for c in l) / len(l) > 0.8
]
print(len(lineas), "líneas")
```

```python
subprocess.run(["say", "-v", voz, "-r", "170", ". ".join(lineas)], check=True)
```
