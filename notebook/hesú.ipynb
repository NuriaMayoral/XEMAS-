```python
import matplotlib
matplotlib.use("Agg")  # backend sin ventanas: evita cuelgues del kernel
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import pandas as pd

df = pd.read_csv(
    "../data/Incendis_forestals_a_Catalunya._Anys_2011-2024_20261116.csv",
    decimal=",", thousands=".",
)
df["HAFORESTAL"] = df["HAFORESTAL"].fillna(0)

# Hectáreas totales y nº de incendios por municipio
muni = (
    df.groupby(["CODI MUNICIPI", "TERMEMUNIC", "COMARCA"])
      .agg(ha=("HAFORESTAL", "sum"), n=("HAFORESTAL", "size"))
      .reset_index()
)
top = muni.nlargest(15, "ha").sort_values("ha")
labels = top["TERMEMUNIC"] + " (" + top["COMARCA"] + ")"

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(labels, top["ha"], color="#B5361E")
for y, (ha, n) in enumerate(zip(top["ha"], top["n"])):
    ax.text(ha, y, f"  {ha:,.0f} ha · {n} incendios", va="center", fontsize=9)
ax.set_title("Top 15 municipios por superficie forestal quemada, 2011–2024", fontweight="bold")
ax.set_xlabel("Hectáreas")
ax.xaxis.set_major_formatter(mtick.StrMethodFormatter("{x:,.0f}"))
ax.set_xlim(0, top["ha"].max() * 1.3)
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout()
fig.savefig("top_municipios.png", dpi=150)
plt.close(fig)
```
